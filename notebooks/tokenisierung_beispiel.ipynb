# Tokenisierung – konkretes Beispiel

Wir betrachten den Eingabetext:

> **„Paris ist die Hauptstadt von“**

Ziel:
- Wortfolge anzeigen
- Tokenfolge (Subword-Token)
- Token-IDs
- Anzahl der Tokens
- Hinweis: *Ein Wort besteht aus mehreren Tokens*

```python
from transformers import AutoTokenizer
import torch

# GPT-2 Tokenizer (passt zu deinem Beispiel mit 768-dimensionalen Embeddings)
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
text = "Paris ist die Hauptstadt von"

# einfache Wortzerlegung (nur zur Illustration!)
word_sequence = text.split()

print("Beispiel-Wortfolge:")
print(word_sequence)
```

```python
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Eingabetext:", text)
print()
print("Token-Strings:", tokens)
print("Token-IDs:", token_ids)
print("Anzahl der Tokens:", len(tokens))
```

### Hinweis

Ein **Wort besteht nicht notwendigerweise aus genau einem Token**.

Beispiel (typisch bei GPT-2 BPE):
- **„Hauptstadt“** → `['ĠHau', 'pt', 'stadt']`
- **„ist“** → `['Ġis', 't']`

Das Vokabular besteht aus **Subword-Tokens**, nicht aus vollständigen Wörtern.

> 💡 Das führende `Ġ` steht für *Leerzeichen vor dem Token* (GPT-2-Tokenizer).


```python
text2 = "Berlin ist die Hauptstadt von Deutschland. Paris ist die Hauptstadt von"

tokens2 = tokenizer.tokenize(text2)
ids2 = tokenizer.convert_tokens_to_ids(tokens2)

print("Eingabetext:", text2)
print()
print("Token-Strings:", tokens2)
print("Token-IDs:", ids2)
print("Anzahl der Tokens:", len(tokens2))
```

```python
# Anzahl Tokens (aus dem ersten Beispiel)
n_tokens = len(tokens)

# GPT-2 Embedding-Dimension
embedding_dim = 768

# Symbolische Embedding-Matrix
X = torch.zeros((n_tokens, embedding_dim))

print(f"Embedding-Matrix X ∈ ℝ^({n_tokens} × {embedding_dim})")
```

### Embeddings

Jeder Token wird auf einen Vektor der Länge **768** abgebildet (GPT-2).

Damit entsteht eine Matrix:

\[
X \in \mathbb{R}^{8 \times 768}
\]

- Zeilen → Tokens
- Spalten → numerische Merkmale (Embedding-Dimensionen)

Diese Matrix ist der **Eingang für den Transformer**.


### Kosten

API-Anbieter berechnen die Kosten meist pro Token (oft in Einheiten von 1.000 oder 1.000.000 Tokens, getrennt für Eingabe und Ausgabe). Mehr Tokens bedeuten also nicht nur mehr Rechenaufwand, sondern auch höhere Kosten – unabhängig vom konkreten Anbieter oder Tarif.

```python
# Illustrative Kostenrechnung (Platzhalter-Preis, kein realer Tarif)
preis_pro_1000_tokens = 0.01  # Platzhaltereinheit – reale Preise variieren je nach Anbieter/Modell

kosten_beispiel1 = (len(tokens) / 1000) * preis_pro_1000_tokens
kosten_beispiel2 = (len(tokens2) / 1000) * preis_pro_1000_tokens

print(f"Beispiel 1: {len(tokens)} Tokens -> ca. {kosten_beispiel1:.5f} (Platzhaltereinheit)")
print(f"Beispiel 2: {len(tokens2)} Tokens -> ca. {kosten_beispiel2:.5f} (Platzhaltereinheit)")
print(f"Faktor: {len(tokens2)/len(tokens):.2f}x mehr Tokens -> {kosten_beispiel2/kosten_beispiel1:.2f}x mehr Kosten")
```
