```python
"""
Katze-vs-Hund-Beispiel für Supervised Learning:

Merkmale: Schnauzenlaenge (x1), Ohrgroesse (x2), jeweils normiert auf [0,1]
Label: y = 0 (Katze) oder y = 1 (Hund) -- von Menschen vergeben

Wir trainieren einen einfachen linearen Klassifikator (logistische Regression)
mit Gradientenabstieg, wie schon in gradient_descent_demo.ipynb:

p = sigmoid(w . x + b)
dL/dw = (p - y) * x
dL/db = (p - y)
w_neu = w_alt - eta * dL/dw
"""

import numpy as np


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def train_step(w, b, x, y, eta):
    p = sigmoid(w @ x + b)
    grad_w = (p - y) * x
    grad_b = (p - y)
    w_new = w - eta * grad_w
    b_new = b - eta * grad_b
    return w_new, b_new, p


def main():
    # Trainingsdaten: (Schnauzenlaenge, Ohrgroesse), Label von Menschen vergeben
    X = np.array([
        [0.2, 0.3],   # Katze
        [0.3, 0.4],   # Katze
        [0.25, 0.35], # Katze
        [0.8, 0.7],   # Hund
        [0.7, 0.6],   # Hund
        [0.9, 0.8],   # Hund
    ])
    y = np.array([0, 0, 0, 1, 1, 1])
    names = ["Katze 1", "Katze 2", "Katze 3", "Hund 1", "Hund 2", "Hund 3"]

    w = np.zeros(2)
    b = 0.0
    eta = 0.5
    epochs = 200

    print("Supervised Learning: Katze-vs-Hund-Klassifikator")
    print("Merkmale: x1 = Schnauzenlaenge, x2 = Ohrgroesse (normiert 0-1)")
    print("Labels (von Menschen vergeben): 0 = Katze, 1 = Hund\n")

    for epoch in range(epochs):
        for xi, yi in zip(X, y):
            w, b, _ = train_step(w, b, xi, yi, eta)

    print(f"Gelernte Gewichte nach {epochs} Epochen: w = {w.round(3)}, b = {b:.3f}\n")

    print("Vorhersagen auf den Trainingsdaten:")
    for name, xi, yi in zip(names, X, y):
        p = sigmoid(w @ xi + b)
        pred = "Hund" if p > 0.5 else "Katze"
        print(f"  {name:8s}: p(Hund) = {p:.3f} -> Vorhersage: {pred} (Label war: {'Hund' if yi==1 else 'Katze'})")

    print("\nNeues, unbekanntes Tier: x = (0.75, 0.65)")
    x_neu = np.array([0.75, 0.65])
    p_neu = sigmoid(w @ x_neu + b)
    pred_neu = "Hund" if p_neu > 0.5 else "Katze"
    print(f"  p(Hund) = {p_neu:.3f} -> Vorhersage: {pred_neu}")

    print("\nInterpretation:")
    print("- Das Modell hat nur aus den gelabelten Beispielen (Katze/Hund) gelernt.")
    print("- Die Labels kamen von Menschen -> das ist Supervised Learning.")
    print("- Anders als beim GPT-Beispiel entstehen die Labels hier NICHT aus den Daten selbst.")
    print("- Gelernt wird mit demselben Gradientenabstieg wie in gradient_descent_demo.ipynb,")
    print("  nur jetzt mit zwei Gewichten (einem pro Merkmal) statt nur einem.")


if __name__ == "__main__":
    main()
```
